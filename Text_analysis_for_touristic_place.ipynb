```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
from pprint import pprint
```

```python
model=SentenceTransformer('paraphrase-MiniLM-L6-v2')
```

```python
sentences=\
    ['Nothing much to say as it is a macbook. the M1 processor works like a chars'
    , 'Amazing laptop, super performance with M1, its blazing fast',
     'Working very slow and takes 15-20 minutes to start thus not worth for money'
     ,
     'This is not a good laptop. It is very slow. It is taking 20 minutes to start'
    ]
```

```python
sentence_embeddings = model.encode(sentences)
```

```python
for sentence, embedding in zip(sentences, sentence_embeddings):
  print("Sentence: ",sentence)
  print("Embedding: ",embedding)
  print("")
```

```python
len(sentence_embeddings)
```

```python
len(sentence_embeddings[0])
```

```python
pprint('Similarity between {} and {} is {}'.format(sentences[0],sentences[1], cosine_similarity(sentence_embeddings[0].reshape(1,-1),sentence_embeddings[1].reshape(1, -1))[0][0]))
```

```python
new_sentences="I think, Macbook is amazing computer for us. Especially with the M4"
```

```python
new_sentence_embeddings=model.encode(new_sentences)
```

```python
similarities=[]
for i, sentence_embedding in enumerate(sentence_embeddings):
  similarity=cosine_similarity(new_sentence_embeddings.reshape(1, -1), sentence_embedding.reshape(1, -1))[0][0]
  similarities.append(similarity)
  print(f"Similarity between new_sentence and sentences [{i}] is {similarity}")
```

```python
max_similarity=max(similarities)
most_similart_index=similarities.index(max_similarity)
```

```python
print(f"New sentence is most similart to sentences [{most_similart_index}] with similarity {max_similarity}")
```

**Buradan sonra SBERT'i denedim. sentence-transformers kütüphanesi kullanılarak SBERT ile cümle benzerliği ölçmeye çalıştım**

```python
from sentence_transformers import SentenceTransformer, util
```

```python
model2=SentenceTransformer("all-MiniLM-L6-v2")
```

```python
description_embeddings_2=model2.encode(data['Description'], convert_to_tensor=True)
```

```python
new_message_2="A tranquil sanctuary offering breathtaking natural beauty, serene walking paths, and a haven of peace away from the bustle."
```

```python
#description_embeddings_2=model2.encode(data['Description'], convert_to_tensor=True)
user_input_embedding=model2.encode(new_message_2, convert_to_tensor=True)
```

```python
similarities_2=util.cos_sim(user_input_embedding,description_embeddings_2)
```

```python
most_similar_idx_2=similarities_2.argmax().item()
most_similar_score=similarities_2[0][most_similar_idx_2]

print(f"User entered text: {new_message}")
print(f"Most similar description: {data['Description'][most_similar_idx_2]}")
print(f"Similarity score: {most_similar_score}")
print(f"İndex: {most_similar_idx_2}")
```
