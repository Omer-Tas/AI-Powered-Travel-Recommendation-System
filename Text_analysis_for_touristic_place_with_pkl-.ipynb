```python
import pandas as pd
import numpy as np
from sentence_transformers import SentenceTransformer, util
import pickle
```

```python
model=SentenceTransformer("all-mpnet-base-v2")
```

```python
data=pd.read_csv("touristic_places_with_reviews.csv")
```

```python
df=data.head(1000)
```

```python
df.shape
```

```python
df['review_embedding']=df['Review'].apply(lambda x:model.encode(x).tolist())
```

```python
df['review_embedding']
```

```python
with open('model_and_data.pkl','wb') as f:
  pickle.dump((model, df), f)
```

**Pkl dosyasını kullanma**

```python
with open('model_and_data.pkl','rb') as f:
  model, df = pickle.load(f)
```

```python
# embedding sütununu nump array olarak alıyoruz

review_embeddings=np.array(df['review_embedding'].tolist())
```

```python
review_embeddings
```

```python
from sklearn.metrics.pairwise import cosine_similarity
import torch
```

```python
new_message="This location seems like a dream for nature lovers. The lush greenery, serene walking paths, and unique wildlife are exactly what I imagine in a perfect getaway. From the pictures, it looks peaceful and full of charm, ideal for relaxation or a refreshing escape from the city life. I'd love to visit someday!"
```

```python
message_embedding=model.encode(new_message, convert_to_tensor=True).float()
```

```python
review_embeddings=torch.tensor(review_embeddings, dtype=torch.float)
```

```python
similarities=util.cos_sim(message_embedding, review_embeddings)
```

```python
most_similar_idx=similarities.argmax().item()
most_similar_score=similarities[0][most_similar_idx]

print(f"User entered text: {new_message}")
print(f"Most similar review {df['Review'][most_similar_idx]}")
print(f"Similarity socre: {most_similar_score}")
```

```python
import pandas as pd
import numpy as np
from sentence_transformers import SentenceTransformer, util
from sklearn.metrics.pairwise import cosine_similarity
import torch
```

```python
model=SentenceTransformer("all-mpnet-base-v2")
```

```python
data=pd.read_pickle("reviews_cleaned.pickle")
```

```python
df=data.head(1000)
```

```python
df.columns
```

```python
df['clean_review'].isna().sum()
```

```python
review_embeddings = model.encode(df['clean_review'].tolist(), convert_to_tensor=True)
```

```python
deneme_message="Amazing vacation! The staff was incredibly friendly, especially Sophia at the front desk. The food was delicious, and the kids loved the activities. The rooms were clean, with a stunning ocean view. Only wish there were more vegetarian options at the buffet. Overall, highly recommend for a relaxing family getaway"
```

```python
deneme_embedding=model.encode(deneme_message, convert_to_tensor=True).float()
```

```python
review_embeddings=torch.tensor(review_embeddings, dtype=torch.float)
```

```python
similarities=util.cos_sim(deneme_embedding, review_embeddings)
```

```python
most_similar_idx=similarities.argmax().item()
most_similar_score=similarities[0][most_similar_idx]

print(f"User entered text: {deneme_message}")
print(f"Most similar review {df['clean_review'][most_similar_idx]}")
print(f"Similarity socre: {most_similar_score}")
```
